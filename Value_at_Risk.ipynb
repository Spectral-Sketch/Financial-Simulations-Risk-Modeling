```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. PARAMETERS
np.random.seed(42)
n_sims = 1000      # We gebruiken veel scenario's voor een mooie grafiek
mu = 0.0005        # Dagelijks rendement
sigma = 0.015      # Dagelijkse volatiliteit
S0 = 100           # Startprijs

# 2. DATA GENEREREN (Eindprijzen na 1 dag)
# We berekenen 1000 mogelijke prijzen voor morgen
daily_returns = np.random.normal(mu, sigma, n_sims)
end_prices = S0 * (1 + daily_returns)
losses = S0 - end_prices # Positief getal is verlies

# 3. GRAFIEK OPZETTEN
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_title("Value at Risk (VaR): Het Risico in Kaart", fontsize=14, fontweight='bold')
ax.set_xlabel("Verlies / Winst in Euro's")
ax.set_ylabel("Hoe vaak komt dit voor?")

# 4. ANIMATIE LOGICA
def update(frame):
    if frame < 50: return # Even wachten tot we genoeg data hebben
    
    ax.clear()
    current_data = losses[:frame]
    
    # Teken het histogram
    n, bins, patches = ax.hist(current_data, bins=30, color='skyblue', edgecolor='white', alpha=0.7)
    
    # Bereken de 95% VaR (de slechtste 5% van de gevallen)
    var_95 = np.percentile(current_data, 95)
    
    # Kleur de 'risico zone' rood
    for i in range(len(patches)):
        if bins[i] > var_95:
            patches[i].set_facecolor('crimson')
            patches[i].set_alpha(0.6)
            
    # Voeg een verticale lijn toe voor de VaR
    ax.axvline(var_95, color='red', lw=2, ls='--')
    ax.set_title(f"VaR Analyse: {frame} scenario's | 95% VaR = {var_95:.2f} euro", fontsize=14, fontweight='bold')
    ax.set_xlabel("Verlies (+) / Winst (-) in Euro's")

# 5. OPSLAAN
print("VaR-animatie wordt gemaakt... dit duurt iets langer door de 1000 scenario's.")
ani = FuncAnimation(fig, update, frames=range(0, n_sims, 20), interval=100)
ani.save('var_uitleg.gif', writer='pillow', fps=10)
print("Klaar! 'var_uitleg.gif' staat in je map.")
plt.close()
```
