```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. INSTELLINGEN
np.random.seed(42)
S0 = 100          # Startprijs
mu = 0.1          # Drift (10% rendement per jaar)
sigma = 0.25      # Volatiliteit (25% beweging)
T = 1.0           # Tijdspanne (1 jaar)
n_steps = 250     # Aantal stapjes (beursdagen)
dt = T / n_steps  # Grootte van elk stapje

# 2. DE WISKUNDE (GBM Formule)
t = np.linspace(0, T, n_steps)
# We berekenen de koersstapjes
W = np.random.standard_normal(size=n_steps) 
W = np.cumsum(W) * np.sqrt(dt) # De 'toevalsfactor'
X = (mu - 0.5 * sigma**2) * t + sigma * W
price_path = S0 * np.exp(X) # De uiteindelijke koers

# 3. DE GRAFIEK OPZETTEN
fig, ax = plt.subplots(figsize=(10, 6))
line, = ax.plot([], [], color='blue', lw=2, label='GBM Pad')
ax.set_xlim(0, T)
ax.set_ylim(min(price_path)*0.9, max(price_path)*1.1)
ax.set_title("Geometric Brownian Motion: De Wiskunde van een Koers", fontsize=14)
ax.set_xlabel("Tijd (Jaren)")
ax.set_ylabel("Prijs")
ax.grid(True, alpha=0.3)

# 4. DE ANIMATIE LOGICA
def update(frame):
    line.set_data(t[:frame], price_path[:frame])
    return line,

# 5. OPSLAAN ALS GIF
print("Simulatie wordt berekend...")
ani = FuncAnimation(fig, update, frames=range(0, n_steps, 2), interval=60, blit=True)
ani.save('gbm_uitleg.gif', writer='pillow', fps=12)
print("Klaar! 'gbm_uitleg.gif' staat in je nieuwe map.")
plt.close()
```
