```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. PARAMETERS (Zelfde als GBM, maar nu voor 50 paden)
np.random.seed(42)
S0 = 100
mu = 0.1
sigma = 0.25
T = 1.0
n_steps = 100     # Iets minder stappen voor een snellere GIF
n_sims = 50       # 50 verschillende scenario's
dt = T / n_steps

# 2. DATA GENEREREN
t = np.linspace(0, T, n_steps)
paths = np.zeros((n_sims, n_steps))

for i in range(n_sims):
    W = np.random.standard_normal(size=n_steps)
    W = np.cumsum(W) * np.sqrt(dt)
    X = (mu - 0.5 * sigma**2) * t + sigma * W
    paths[i] = S0 * np.exp(X)

# 3. GRAFIEK OPZETTEN
fig, ax = plt.subplots(figsize=(10, 6))
lines = [ax.plot([], [], lw=1, alpha=0.3, color='blue')[0] for _ in range(n_sims)]
mean_line, = ax.plot([], [], color='red', lw=3, label='Gemiddelde Verwachting')

ax.set_xlim(0, T)
ax.set_ylim(40, 200) # Ruime marges voor alle scenario's
ax.set_title("Monte Carlo: 50 Mogelijke Toekomstscenario's", fontsize=14, fontweight='bold')
ax.set_xlabel("Tijd (Jaren)")
ax.set_ylabel("Prijs")
ax.legend(loc='upper left')

# 4. ANIMATIE LOGICA
def update(frame):
    for i in range(n_sims):
        lines[i].set_data(t[:frame], paths[i, :frame])
    
    # Bereken het gemiddelde van alle paden tot dat moment
    avg_path = np.mean(paths[:, :frame], axis=0)
    mean_line.set_data(t[:frame], avg_path)
    return lines + [mean_line]

# 5. OPSLAAN
print("Monte Carlo simulatie wordt gemaakt...")
ani = FuncAnimation(fig, update, frames=range(0, n_steps, 2), interval=80, blit=True)
ani.save('monte_carlo_uitleg.gif', writer='pillow', fps=12)
print("Klaar! 'monte_carlo_uitleg.gif' is opgeslagen.")
plt.close()
```
